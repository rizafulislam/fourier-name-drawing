# Fourier name drawing

Draws one or two lines of text (Bangla or English) with a Fourier series of rotating circles.
Edit the names in the second cell, then choose **Runtime → Run all**. Rendering takes about a minute.

```python
from fourier_name import make_video, Style

lines = ["রিজাফুল", "RIZAFUL"]   # one or two lines, Bangla or English
terms = 800                      # more terms give sharper letters

make_video(lines, "name.mp4", terms=terms, style=Style(draw=24, hold=4))
```

```python
from base64 import b64encode
from IPython.display import HTML
from google.colab import files

data = b64encode(open("name.mp4", "rb").read()).decode()
display(HTML(f'<video width="720" controls autoplay loop src="data:video/mp4;base64,{data}"></video>'))
files.download("name.mp4")
```
